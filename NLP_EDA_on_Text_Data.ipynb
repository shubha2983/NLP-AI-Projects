```python
# 1 Read the text data
#Importing data
import nltk
from nltk.corpus import webtext
nltk.download('webtext')
wt_sentences = webtext.sents('firefox.txt')
wt_words = webtext.words('firefox.txt')
```

```python
# 2 Import necessary libraries
from nltk.probability import FreqDist
from nltk.corpus import stopwords
import string
```

```python
# 3 Check number of words in the data 
len(wt_sentences)


```

```python
len(wt_words)
```

```python
# 4 Compute the frequency of all words in the reviews
frequency_dist = nltk.FreqDist(wt_words)
frequency_dist
```

```python
sorted_frequency_dist =sorted(frequency_dist,key=frequency_dist.__getitem__, reverse=True)

sorted_frequency_dist
```

```python
# 5 Consider words with length greater than 3 and plot
large_words = dict([(k,v) for k,v in frequency_dist.items() if
len(k)>3])
frequency_dist = nltk.FreqDist(large_words)
frequency_dist.plot(50,cumulative=False)
```
