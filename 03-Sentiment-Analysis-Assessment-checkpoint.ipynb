## Task #1: Perform vector arithmetic on your own words
Write code that evaluates vector arithmetic on your own set of related words. The goal is to come as close to an expected word as possible. Please feel free to share success stories in the Q&A Forum for this section!

```python
# Import spaCy and load the language library. Remember to use a larger model!

import spacy
```

```python
nlp=spacy.load('en_core_web_lg')
```

```python
# Choose the words you wish to compare, and obtain their vectors

Ball=nlp(u"Ball")
Base=nlp(u"Base")
```

```python
Ball.vector
```

```python
Base.vector
```

```python
# Import spatial and define a cosine_similarity function
from scipy import spatial 

cosine_similarity=lambda vec1,vec2 : 1-spatial.distance.cosine(vec1,vec2)

```

```python
cosine_similarity(Ball.vector,Ball.vector)
```

```python
Cricket=nlp(u"Cricket")
```

```python
# Write an expression for vector arithmetic
# For example: new_vector = word1 - word2 + word3
new_vector= Ball.vector - Base.vector + Cricket.vector
```

```python
len(nlp.vocab)
```

```python
# List the top ten closest vectors in the vocabulary to the result of the expression above
computed_similarites=[]

for word in nlp.vocab:
    if word.is_lower:
        if word.has_vector:
            if word.is_alpha:
                similarity=cosine_similarity(new_vector,word.vector)
                computed_similarites.append((word.text,similarity))
            










```

```python
computed_similarites=sorted(computed_similarites,key= lambda item : -item[1] )
```
